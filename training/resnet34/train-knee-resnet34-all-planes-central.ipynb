# ResNet34 single-model variation: all planes and central sampling

Variation of the original four-triplet ResNet34 experiment whose resumed model achieved public score 0.839 (submission V9). Uses the original seed-42, 20% grouped holdout, original 3,917-study image bundle, community soft labels and preserved official labels. With the same inputs this is 3,133 training studies, 784 held-out studies, and 9 officially labelled validation studies. This is one model, not the repeated-holdout ensemble. No member index, saved ensemble split plan or separate 100-study selection set is needed.

Changes: use every usable series in training and evaluation to cover every available anatomical plane; sample four triplet centers from depth regions 0–35%, 35–50%, 50–65%, and 65–100%; shift boundary triplets inward to avoid repeated slices when at least three real slices exist. Two centers emphasize the central 30%, while outer regions remain covered. Training samples randomly within each bin; validation/inference use fixed bin midpoints. These fractions are experimental choices. Short stacks use adjusted nonempty bins, or all centers for at most four real slices. Triplets may overlap. Only stacks shorter than three real slices repeat within a triplet. Never sample depth padding.

Attach the original normalized image dataset, original competition data and nartaa/rsna-knee-hpo-assets community labels. No renormalization is required. This variation uses the original architecture, mean/max pooling, loss, optimizer, patience and 7.5-hour session budget. All-series processing can increase runtime and memory. Retained neighboring slices may not be consecutive acquired slices in long series subsampled during preprocessing.

Start fresh from ImageNet, not the previous 0.839 model's checkpoint. Resume only this variation's own trusted last.pt by setting RESUME_CHECKPOINT; its configuration/source identity must match. Save outputs privately after each session. The output directory and model variation are separate from the original model. Upload defaults to on when this notebook is run; creating the notebook does not train or upload anything.

The exported runtime and manifest retain the new sampling contract. Existing competition submission notebooks reject this contract and must be adapted before submission. Original training and inference notebooks remain unchanged. Because series selection and slice sampling change together, this experiment does not isolate their individual effects.

## Train the CNN and classifier
Community probabilities stay floating point and use masked binary cross-entropy with logits; missing values contribute no loss. Official validation targets are neither smoothed nor replaced. Adam uses 1e-5 for the pretrained CNN and 1e-4 for the head. Validation uses deterministic slices and all usable series. Each epoch scans all training studies.
