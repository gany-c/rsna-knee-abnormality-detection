# ResNet34 — three repeated 25% golden holdouts

Train three separate four-triplet models. This is **repeated holdout**, not disjoint three-fold cross-validation. Each member holds out approximately 25% of golden patient/study groups (15 of 58 studies when every study is its own group). Holdouts can overlap; some golden studies will never be validated. Only held-out predictions qualify for evaluation; do not average predictions from models trained on a study into its validation score.

Attach the original `gany24558/rsna-knee-normalized-training-data` image bundle, competition data, and `nartaa/rsna-knee-hpo-assets` community labels. Enable GPU. Internet is needed for initial ImageNet weights and optional upload; attach pretrained weights for offline initialization. No new normalization is required.

**One member per Kaggle run:** set MEMBER_INDEX to 0, 1, or 2. Keep all split settings fixed between members. Each session has a 7.5-hour training budget; save outputs and resume the same member with its trusted last.pt until early stopping or the epoch cap. A restart without the checkpoint repeats training. Validation and selection export have separate bounded post-training time allowances.

A fixed random set of 100 non-golden studies with usable community labels is excluded from every member's training, including related patient groups. It is reserved for ensemble selection, not an independent test. There are no radiologist labels yet. With three members, there is only one three-model combination; these predictions can compare individuals, pairs, or the full trio. Soft-target loss measures agreement with community supervision; AUROC should not treat soft probabilities as verified binary labels.

The three exported model variations are `comm-res-rh25-member-0`, `comm-res-rh25-member-1`, and `comm-res-rh25-member-2`. Upload defaults to on and includes best-so-far checkpoints after time-budget stops. Upload notes distinguish incomplete from completed training; resume from the private last.pt, not the inference package. This notebook does not build or submit ensemble inference; benchmark that separately against the nine-hour submission limit.

```python
BEST,HISTORY,STOP_REASON=cnn.train_model(MODEL,TRAIN,VAL,SERIES,INDEX['identity']['run_id'],CFG,WORK,IDENTITY,RESUME_CHECKPOINT)
print('Member:',MEMBER_INDEX,'Stop reason:',STOP_REASON,'Best epoch:',BEST['epoch'])
display(pd.DataFrame(HISTORY))
FINISHED = STOP_REASON in ('early_stopping','max_epochs')
if not FINISHED: print('Save these outputs privately and resume this same member from last.pt. Do not switch to another member expecting continuation.')
```

## Export the best checkpoint

```python
PACKAGE=cnn.export_model(MODEL,BEST,CFG,INDEX,IDENTITY,WORK,SOURCE_FILES)
print('Exported and reload-tested:',PACKAGE)
# Reconstruct predictions from the best state even when it came from an attached checkpoint.
import time
by_study={uid:g.sort_values(data.SID).to_dict('records') for uid,g in SERIES.groupby(data.ID)}
with torch.inference_mode():
    loss,metrics,pred=cnn.validate(MODEL,VAL,by_study,INDEX['identity']['run_id'],CFG,DEVICE,time.monotonic()+1800)
frame=pd.DataFrame(pred,columns=data.TARGETS)
frame.insert(0,data.ID,VAL[data.ID].tolist())
frame.to_csv(WORK/'validation_predictions_private.csv',index=False)
print('Best checkpoint validation:',loss,metrics)
```

## Optional final selection predictions
These use community soft targets, not verified medical truth. They never select training checkpoints. Run only after this member has finished training.

```python
if PREDICT_SELECTION and FINISHED and len(SELECTION):
    outputs=[]; MODEL.eval(); deadline=time.monotonic()+1800
    with torch.inference_mode():
        for row in SELECTION.to_dict('records'):
            cnn.check_budget(deadline)
            images=cnn.study_images(row[data.ID],by_study,INDEX['identity']['run_id'],CFG)
            with torch.autocast(DEVICE.type,dtype=torch.float16,enabled=DEVICE.type=='cuda'):
                z=MODEL(images,CFG['microbatch'])
            outputs.append(z.float().sigmoid().cpu().numpy()[0])
    probabilities=np.asarray(outputs)
    if not np.isfinite(probabilities).all():raise ValueError('Nonfinite selection predictions')
    frame=pd.DataFrame(probabilities,columns=data.TARGETS)
    frame.insert(0,data.ID,SELECTION[data.ID].tolist())
    frame.to_csv(WORK/'selection_predictions_private.csv',index=False)
    SELECTION[[data.ID]+data.TARGETS+[t+'__mask' for t in data.TARGETS]].to_csv(WORK/'selection_targets_private.csv',index=False)
    y=SELECTION[data.TARGETS].fillna(0).to_numpy(float)
    mask=SELECTION[[t+'__mask' for t in data.TARGETS]].to_numpy(bool)
    p=np.clip(probabilities,1e-7,1-1e-7)
    bce=float((-(y*np.log(p)+(1-y)*np.log(1-p)))[mask].mean())
    data.atomic_json(WORK/'selection_metrics_private.json',dict(soft_target_bce=bce,studies=len(SELECTION),member=MEMBER_INDEX))
    print('Community selection loss (not medical accuracy):',bce)
else:
    print('Selection predictions skipped: enabled only for completed members with a nonempty selection set.')
```
